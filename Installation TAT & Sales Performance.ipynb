```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Registration Report.csv", parse_dates=["CREATION DATE", "INSTALLATION DATE"])

# --- Feature Engineering ---
# Calculate TAT (Turnaround Time) in days
df["TAT_days"] = (df["INSTALLATION DATE"] - df["CREATION DATE"]).dt.days

# Categorize SLA buckets
def sla_bucket(x):
    if pd.isna(x):
        return "Missing"
    elif x <= 2:
        return "0-2 days"
    elif x <= 5:
        return "2-5 days"
    elif x <= 7:
        return "5-7 days"
    else:
        return "Above 7 days"

df["SLA_Category"] = df["TAT_days"].apply(sla_bucket)

# Extract month/year for grouping
df["Month"] = df["CREATION DATE"].dt.to_period("M")

# --- Executive Productivity ---
exec_prod = df.groupby(["Sales Executive Name", "Month"]).agg(
    Customers=("Account no", "count"),
    Total_Value=("Plan Value in rupees", "sum")
).reset_index()

# Yearly totals per executive
exec_year = df.groupby("Sales Executive Name").agg(
    Customers=("Account no", "count"),
    Total_Value=("Plan Value in rupees", "sum")
).reset_index()

# --- Sales Manager Productivity ---
mgr_prod = df.groupby(["Sales Manager", "Month"]).agg(
    Customers=("Account no", "count"),
    Total_Value=("Plan Value in rupees", "sum")
).reset_index()

mgr_year = df.groupby("Sales Manager").agg(
    Customers=("Account no", "count"),
    Total_Value=("Plan Value in rupees", "sum")
).reset_index()

# --- Node with Maximum Sales per City ---
node_city = df.groupby(["City", "Installation node"]).agg(
    Customers=("Account no", "count")
).reset_index()

node_max = node_city.loc[node_city.groupby("City")["Customers"].idxmax()]

# --- Speed & Validity Patterns ---
speed_validity = df.groupby(["Month", "Plan speeds", "Validity"]).agg(
    Customers=("Account no", "count"),
    Total_Value=("Plan Value in rupees", "sum")
).reset_index()

# --- SLA Distribution ---
sla_dist = df.groupby(["Month", "SLA_Category"]).agg(
    Customers=("Account no", "count")
).reset_index()

# --- Outlier Analysis ---
outliers = df[df["TAT_days"] > 7]

# --- Save outputs ---
exec_prod.to_csv("executive_monthly.csv", index=False)
exec_year.to_csv("executive_yearly.csv", index=False)
mgr_prod.to_csv("manager_monthly.csv", index=False)
mgr_year.to_csv("manager_yearly.csv", index=False)
node_max.to_csv("node_max_city.csv", index=False)
speed_validity.to_csv("speed_validity_patterns.csv", index=False)
sla_dist.to_csv("sla_distribution.csv", index=False)
outliers.to_csv("tat_outliers.csv", index=False)

print("Analysis complete. CSVs saved.")
```

```python
# --- Executive Summary Generator Testing Notebook ---

import pandas as pd

# Universal formatter
def format_value(num):
    if num >= 1_000_000:
        return f"{num/1_000_000:.1f} M"
    elif num >= 100_000:
        return f"{num/100_000:.1f} L"
    else:
        return f"{round(num):,}"

# Enhanced summary generator
def generate_summary(df, level="Month", group_col=None, value_col=None, install_col=None):
    summary = ""

    if level == "Month":
        last = df.iloc[-1]
        prev = df.iloc[-2] if len(df) > 1 else None
        if prev is not None:
            change_installs = last[install_col] - prev[install_col]
            change_value = last[value_col] - prev[value_col]

            pct_installs = (change_installs / prev[install_col] * 100) if prev[install_col] != 0 else 0
            pct_value = (change_value / prev[value_col] * 100) if prev[value_col] != 0 else 0

            summary = (
                f"In {last['MonthYear']}, installs were {last[install_col]} "
                f"({change_installs:+}, {pct_installs:+.1f}% vs {prev['MonthYear']}), "
                f"and value was ₹{format_value(last[value_col])} "
                f"({format_value(change_value)} change, {pct_value:+.1f}%)."
            )
        else:
            summary = f"Initial month {last['MonthYear']} shows {last[install_col]} installs and ₹{format_value(last[value_col])} value."

    elif level in ["City","Executive","Node","PlanValidity","Speed"]:
        grouped = df.groupby(group_col).agg(
            Installs=(install_col,"sum"),
            Value=(value_col,"sum")
        ).reset_index()

        top3 = grouped.sort_values("Installs",ascending=False).head(3)
        top_summary = "; ".join([
            f"{row[group_col]} ({row['Installs']} installs, ₹{format_value(row['Value'])})"
            for _, row in top3.iterrows()
        ])

        bottom = grouped.sort_values("Installs",ascending=True).iloc[0]

        total_installs = grouped["Installs"].sum()
        total_value = grouped["Value"].sum()

        summary = (
            f"Top 3 {level}s by installs: {top_summary}. "
            f"Lowest {level}: {bottom[group_col]} ({bottom['Installs']} installs, ₹{format_value(bottom['Value'])}). "
            f"Overall {level}-level total: {total_installs} installs, ₹{format_value(total_value)}."
        )

    return summary
```

```python
import pandas as pd

# Load your New Registration Report
new_reg_df = pd.read_csv("New Registration Report.csv")

# --- Universal Formatter ---
def format_value(num):
    if num >= 1_000_000:
        return f"{num/1_000_000:.1f} M"
    elif num >= 100_000:
        return f"{num/100_000:.1f} L"
    else:
        return f"{round(num):,}"

# --- Column Mapping Helper ---
def map_columns(df):
    col_map = {}
    for col in df.columns:
        low = col.lower()
        if "account" in low or "acct" in low:
            col_map["install_col"] = col
        elif "plan" in low and "val" in low:
            col_map["value_col"] = col
        elif "exec" in low:
            col_map["exec_col"] = col
        elif "city" in low:
            col_map["city_col"] = col
        elif "node" in low:
            col_map["node_col"] = col
        elif "valid" in low:
            col_map["validity_col"] = col
        elif "speed" in low:
            col_map["speed_col"] = col
        elif "date" in low:
            col_map["date_col"] = col
    return col_map

# --- Trendline Streak Detector ---
def detect_streak(series, label="installs"):
    streak = 0
    direction = None
    for i in range(1, len(series)):
        if series[i] > series[i-1]:
            if direction == "up":
                streak += 1
            else:
                direction = "up"
                streak = 1
        elif series[i] < series[i-1]:
            if direction == "down":
                streak += 1
            else:
                direction = "down"
                streak = 1
    if streak >= 2:
        return f"{label.capitalize()} have {direction} for {streak} consecutive periods."
    return ""

# --- Enhanced Summary Generator ---
def generate_summary(df, level="Month"):
    cols = map_columns(df)
    install_col = cols.get("install_col")
    value_col = cols.get("value_col")

    # Auto-create MonthYear if missing
    if level == "Month":
        if "MonthYear" not in df.columns:
            if "date_col" in cols:
                df[cols["date_col"]] = pd.to_datetime(df[cols["date_col"]], errors="coerce")
                df["MonthYear"] = df[cols["date_col"]].dt.strftime("%b-%Y")
            else:
                raise KeyError("No date column found to create MonthYear")

        grouped = df.groupby("MonthYear").agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

        last = grouped.iloc[-1]
        prev = grouped.iloc[-2] if len(grouped) > 1 else None

        if prev is not None:
            change_installs = last["Installs"] - prev["Installs"]
            change_value = last["Value"] - prev["Value"]

            pct_installs = (change_installs / prev["Installs"] * 100) if prev["Installs"] != 0 else 0
            pct_value = (change_value / prev["Value"] * 100) if prev["Value"] != 0 else 0

            streak_msg = detect_streak(grouped["Installs"].tolist(), "installs")
            streak_val_msg = detect_streak(grouped["Value"].tolist(), "value")

            return (
                f"In {last['MonthYear']}, installs were {last['Installs']} "
                f"({change_installs:+}, {pct_installs:+.1f}% vs {prev['MonthYear']}), "
                f"and value was ₹{format_value(last['Value'])} "
                f"({format_value(change_value)} change, {pct_value:+.1f}%). "
                f"{streak_msg} {streak_val_msg}"
            )
        else:
            return f"Initial month {last['MonthYear']} shows {last['Installs']} installs and ₹{format_value(last['Value'])} value."

    elif level == "City":
        grouped = df.groupby(cols["city_col"]).agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

    elif level == "Executive":
        grouped = df.groupby(cols["exec_col"]).agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

    elif level == "Node":
        grouped = df.groupby(cols["node_col"]).agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

    elif level == "PlanValidity":
        grouped = df.groupby(cols["validity_col"]).agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

    elif level == "Speed":
        grouped = df.groupby(cols["speed_col"]).agg(
            Installs=(install_col,"count"),
            Value=(value_col,"sum")
        ).reset_index()

    # Top 3 + lowest summary
    top3 = grouped.sort_values("Installs",ascending=False).head(3)
    top_summary = "; ".join([
        f"{row.iloc[0]} ({row['Installs']} installs, ₹{format_value(row['Value'])})"
        for _, row in top3.iterrows()
    ])
    bottom = grouped.sort_values("Installs",ascending=True).iloc[0]
    total_installs = grouped["Installs"].sum()
    total_value = grouped["Value"].sum()

    return (
        f"Top 3 {level}s by installs: {top_summary}. "
        f"Lowest {level}: {bottom.iloc[0]} ({bottom['Installs']} installs, ₹{format_value(bottom['Value'])}). "
        f"Overall {level}-level total: {total_installs} installs, ₹{format_value(total_value)}."
    )
```
